### Question 10
Write a function that will convert carat to grams and find the heaviest diamond.

1 carat = 0.2 grams

````python
# function to convert carats to grams
def carat_to_grams(ct):
    return ct * 0.2

print(carat_to_grams(5))
````

````python
# heaviest diamond
diamonds["grams"] = carat_to_grams(diamonds["carat"])
print(diamonds["grams"].max())
print(diamonds[diamonds["grams"] == diamonds["grams"].max()])
````
